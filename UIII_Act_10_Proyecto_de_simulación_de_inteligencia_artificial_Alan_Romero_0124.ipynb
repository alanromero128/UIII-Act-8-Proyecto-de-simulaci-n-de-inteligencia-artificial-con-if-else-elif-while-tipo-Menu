### Lista De modelos

```python
# IA SIMPLE DE MODELOS
# Lista de diccionarios
# Ejemplo para preparatoria

modelo = []       # Diccionario vacío
lasmodelos = {}   # Lista vacía

while True:

    print("\n==============================")
    print("   IA DE MODELOS VIRTUALES")
    print("==============================")
    print("1. Registrar modelo")
    print("2. Ver modelos")
    print("3. Hablar con la IA")
    print("4. Salir")

    opcion = input("Selecciona una opción: ")

    # OPCIÓN 1
    if opcion == "1":

        modelo = {}

        nombre = input("Nombre de la modelo: ")
        edad = input("Edad: ")
        color = input("Color favorito: ")

        modelo["nombre"] = nombre
        modelo["edad"] = edad
        modelo["color"] = color

        lasmodelos.append(modelo)

        print("Modelo registrada correctamente.")

    # OPCIÓN 2
    elif opcion == "2":

        if len(lasmodelos) == 0:
            print("No hay modelos registradas.")

        else:

            print("\n========================================")
            print("           LISTA DE MODELOS")
            print("========================================")
            print("NOMBRE\t\tEDAD\tCOLOR")
            print("----------------------------------------")

            for m in lasmodelos:

                print(m["nombre"], "\t\t", m["edad"], "\t", m["color"])

    # OPCIÓN 3
    elif opcion == "3":

        if len(lasmodelos) == 0:
            print("Primero registra una modelo.")

        else:

            buscar = input("Escribe el nombre de la modelo: ")
            encontrada = False

            for m in lasmodelos:

                if m["nombre"].lower() == buscar.lower():

                    encontrada = True

                    mensaje = input("Habla con la IA: ")

                    if "hola" in mensaje.lower():
                        print(m["nombre"], "dice: Hola 😊")

                    elif "color" in mensaje.lower():
                        print(m["nombre"], "dice: Mi color favorito es", m["color"])

                    elif "edad" in mensaje.lower():
                        print(m["nombre"], "dice: Tengo", m["edad"], "años")

                    else:
                        print(m["nombre"], "dice: Estoy aprendiendo nuevas respuestas.")

            if encontrada == False:
                print("Modelo no encontrada.")

    # OPCIÓN 4
    elif opcion == "4":

        print("Programa finalizado.")
        break

    # ERROR
    else:
        print("Opción inválida.")
```
