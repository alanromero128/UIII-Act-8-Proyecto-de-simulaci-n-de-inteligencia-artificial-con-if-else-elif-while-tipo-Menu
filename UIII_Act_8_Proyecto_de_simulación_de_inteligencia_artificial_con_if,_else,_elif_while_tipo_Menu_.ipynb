```python
print("--- Simulador: El Teclado que Habla ---")
continuar = "si"

while continuar == "si":
    print("\n¿Qué quieres hacer hoy, programador?")
    print("1. Saludar al teclado")
    print("2. Pedir consejo de código")
    print("3. Quejarse del bug")
    print("4. Salir")

    opcion = input("Elige una opción (1-4): ")

    if opcion == "1":
        print("Teclado: ¡Hola! Mis teclas están listas para el código.")
    elif opcion == "2":
        print("Teclado: Si el código no corre, ¡revisa los paréntesis!")
    elif opcion == "3":
        print("Teclado: No llores, seguro es solo un punto y coma... ¡Ah no, esto es Python!")
    elif opcion == "4":
        continuar = "no"
        print("Teclado: ¡Adiós! No me golpees tan fuerte al teclear.")
    else:
        print("Opción no válida, intenta de nuevo.")

print("Programa finalizado.")
```
